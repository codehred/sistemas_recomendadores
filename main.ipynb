```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import linear_kernel, cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer
from ast import literal_eval
import numpy as np
```

```python
import pandas as pd
df = pd.read_csv('./data/movies_metadata.csv', low_memory=False)
df.head(3)
```

```python
C = df['vote_average'].mean()
```

```python
m = df['vote_count'].quantile(0.90)
print(m)
```

```python
q_movies = df.copy().loc[df['vote_count'] >= m]
```

```python
# media ponderada
def weighted_rating(x, m=m, C=C):
    v = x['vote_count']
    R = x['vote_average']

    # Cálculo de IMDB
    return (v/(v+m) * R) + (m/(m+v) * C)
```

```python
q_movies['score'] = q_movies.apply(weighted_rating, axis=1)
```

```python
q_movies = q_movies.sort_values('score', ascending=False)

#Mostrar los primeros 15 resultados
q_movies[['title', 'vote_count', 'vote_average', 'score']].head(15)
```

```python
#Importar TfIdfVectorizer de scikit-learn
from sklearn.feature_extraction.text import TfidfVectorizer

#Definir el objeto de la clase TF-IDF Vectorizer. Quitamos stop words de inglés
tfidf = TfidfVectorizer(stop_words='english')

#Reemplazar NaN por string vacío
df['overview'] = df['overview'].fillna('')

# Consruir la matriz TF-IDF haciendo ajustes y transformaciones
tfidf_matrix = tfidf.fit_transform(df['overview'])

#Mostrar shape
tfidf_matrix.shape
```

```python
# Crea un mapeo de títulos de películas a índices del DataFrame
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
def get_recommendations(title, tfidf_matrix=tfidf_matrix, df=df):
    # 1. Obtenemos el índice de la película por su nombre
    idx = indices[title]

    # 2. En lugar de usar la matriz gigante, calculamos la similitud 
    # SOLO de esta película contra las demás en este momento.
    # Esto es lo que evita que se te cierre el WSL.
    sim_scores_row = linear_kernel(tfidf_matrix[idx], tfidf_matrix).flatten()

    # 3. Lo demás es igual: enumerar y ordenar
    sim_scores = list(enumerate(sim_scores_row))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # 4. Tomamos las 10 mejores (sin contar la misma película)
    sim_scores = sim_scores[1:11]
    movie_indices = [i[0] for i in sim_scores]

    # 5. Regresamos los títulos
    return df['title'].iloc[movie_indices]
```

```python
# Carga de datos
df = pd.read_csv('./data/movies_metadata.csv', low_memory=False)
credits = pd.read_csv('./data/credits.csv')
keywords = pd.read_csv('./data/keywords.csv')

# Limpieza de IDs
df = df.drop([19730, 29503, 35587, 35803])

# Conversión a int
keywords['id'] = keywords['id'].astype('int')
credits['id'] = credits['id'].astype('int')
df['id'] = df['id'].astype('int')

# Merges (Ahora tu df es más completo, tiene actores, keywords, etc.)
df = df.merge(credits, on='id')
df = df.merge(keywords, on='id')

# MUY IMPORTANTE: Después del merge, resetea los índices para que no falten números
df = df.reset_index(drop=True)
```

```python
from ast import literal_eval

features = ['cast', 'crew', 'keywords', 'genres']

for feature in features:
    # Usamos un try/except interno o validamos que sea string antes de evaluar
    # Esto evita que el código truene si encuentra un valor que ya es objeto o es NaN
    df[feature] = df[feature].apply(lambda x: literal_eval(x) if isinstance(x, str) else x)
```

```python
def get_director(x):
    for i in x:
        if i['job'] == 'Director':
            return i['name']
    return np.nan
```

```python
def get_list(x):
    if isinstance(x, list):
        names = [i['name'] for i in x]
        #Checar si existen más de 3 elementos. Si sí, regresar primeros 3, si no, todos
        if len(names) > 3:
            names = names[:3]
        return names

    # regresar lista vacía si los datos no están bien formateados
    return []
```

```python
# Extraer director de columnas crew
df['director'] = df['crew'].apply(get_director)

# Extraer top 3 de elenco, palabras clave y géneros
features = ['cast', 'keywords', 'genres']
for feature in features:
    df[feature] = df[feature].apply(get_list)
```

```python
df[['title', 'cast', 'director', 'keywords', 'genres']].head(3)
```

```python
def clean_data(x):
    if isinstance(x, list):
        return [str.lower(i.replace(" ", "")) for i in x]
    else:
        #Checar si existse el director. Si no, regresar ""
        if isinstance(x, str):
            return str.lower(x.replace(" ", ""))
        else:
            return ''
```

```python
features = ['cast', 'keywords', 'director', 'genres']

for feature in features:
    df[feature] = df[feature].apply(clean_data)
```

```python
def create_soup(x):
    return ' '.join(x['keywords']) + ' ' + ' '.join(x['cast']) + ' ' + x['director'] + ' ' + ' '.join(x['genres'])

df['soup'] = df.apply(create_soup, axis=1)

df[['soup']].head(2)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count = CountVectorizer(stop_words='english')
count_matrix = count.fit_transform(df['soup'])


count_matrix.shape
```

```python
# Resetear el índice es bueno, pero asegúrate de no crear columnas 'index' duplicadas
df = df.reset_index(drop=True) 

# Crear el mapeo de títulos a índices (esto es rápido y ligero)
indices = pd.Series(df.index, index=df['title']).drop_duplicates()

from sklearn.metrics.pairwise import cosine_similarity

def get_recommendations_v2(title, count_matrix=count_matrix, df=df):
    # 1. Obtener el índice de la película
    try:
        idx = indices[title]
    except KeyError:
        return "Película no encontrada en el catálogo."

    # 2. EL TRUCO ANTI-CRASH:
    # Calculamos la similitud solo para esta película en el momento.
    # Usamos count_matrix[idx] contra toda la count_matrix.
    sim_scores_row = cosine_similarity(count_matrix[idx], count_matrix).flatten()

    # 3. Enumerar y ordenar
    sim_scores = list(enumerate(sim_scores_row))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # 4. Top 10 (sin contarse a sí misma)
    sim_scores = sim_scores[1:11]
    movie_indices = [i[0] for i in sim_scores]

    # 5. Retornar títulos
    return df['title'].iloc[movie_indices]
```
